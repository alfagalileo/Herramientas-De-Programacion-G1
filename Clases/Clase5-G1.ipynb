# Bucles
Son instrucciones que permiten ejecutar tareas repetitivas sin usar la técnica de recursión. Estas permiten controlar de manera precisa el número de repeticiones

# `while`
Mientras que una condición lógica se cumpla, una tarea dentro del cuerpo de la instrucción `while` se ejecuta

```python
while #condicion:
    #tarea repetitiva
```

````python
# variable conteo
n = 0

while n <= 50:
    print(n)
    n = n + 1
````

````python
print(n)
````

````python
# variable de conteo
n = 50

while n >= 0:
    print(n)
    n = n - 1
````

**Ej**: Serie de Leibniz para aproximar $\pi$

$$
    \frac{\pi}{4} = \sum_{n=0}^{\infty} \frac{ (-1)^n }{ 2n + 1}
$$

$
   \frac{(-1)^0}{2\times 0 + 1} +  \frac{(-1)^1}{2\times 1 + 1} + \frac{(-1)^2}{2\times 2 + 1}
$


````python
# variable de conteo
n = 0

val = 0.0

while n <= 100_000:
    val = val + (-1)**n/(2*n+1)
    n = n + 1

print(f"El valor aproximado de pi es {4*val}")
````

### Notación corta de modificación en sitio de variables

- `n = n + 1` ➡ `n += 1`
- `n = n - 1` ➡ `n -= 1`

````python
# variable de conteo
n = 100_000

val = 0.0

while n >= 0:
    val = val + (-1)**n/(2*n+1)
    n -= 1

print(f"El valor aproximado de pi es {4*val}")
````

# Bucle `for in`

Esta instrucción genera un conjunto de datos iterables auxiliar para iterar sobre cada uno de ellos

```python
for var in iterable:
    #tarea repetitiva
```

````python
for t in "*"*3:
    print("hola"+t)
````

## Función `range`
Esta función genera un rango de numeros naturales iniciando desde un un valor  `start` hasta un `stop`, dando pasos de tamaño `step`.

```python
range(10) # genera los numeros 0,1,2,...,9
range(1,10) # genera los numeros 1,2,3,...,9
range(1,10,2) # genera los numeros 1,3,5,...,9
```

````python
for v in range(10):
    print(v)
````

````python
for v in range(1,10):
    print(v)
````

````python
for v in range(1,10,2):
    print(v)
````

````python
for n in range(20,0,-1):
    print(n)
````

````python
val = 0.0
for n in range(100_000):
    val += (-1)**n/(2*n+1)
print(f"El valor approximado de pi es {4*val}")
````

## Instrucción `break`


````python
while True:
    print("Hola")
    break
````

**Ej**: Adivinar un número

````python
n = 7

while True:
    # solicitar el valor del usuario
    v = int( input(">> Ingrese un numero entero entre [0,100]: ") )

    if v == n:
        print("Adivinaste!")
        break
    else:
        print("Intentelo de nuevo")

````

````python
def draw_square( L: int ):
    n = 0
    while n < 4:
        forward(L)
        right(90)
        n += 1

speed(10)
initializeTurtle()
draw_square(100)
````

````python
def draw_square2( L: int ):
    for n in range(4):
        forward(L)
        right(90)
        n += 1

speed(10)
initializeTurtle()
draw_square2(100)
````

````python
def make_polygon( L, n ):
    angulo_externo = 360/n

    for j in range(n):
        forward(L)
        right(angulo_externo)

initializeTurtle()
make_polygon( 100, 5 )
````

````python
initializeTurtle()
speed(10)

for j in range(3, 12):
    if j%2 == 0:
        color("yellow")
    else:
        color("blue")
    make_polygon(100, j)
````

````python
initializeTurtle()
speed(10)


for i in range(4):
    for j in range(4):
        penup()
        goto(100 + i*100, 100 + j*100 )
        pendown()
        make_polygon(50, 6)
````
