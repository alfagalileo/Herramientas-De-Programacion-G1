# Manejo de DataFrames con _Pandas_

Un DataFrame es una estructura de datos que permite almacenar información en memoria de forma eficiente y similar a una tabla, ofreciendo diversos métodos para consultar la información de forma simple.

```python
import pandas as pd
```

Creación de DataFrames

```python
notas = pd.DataFrame(
    {
        "Matematicas": [3.5, 4.1, 4.2, 3.2],
        "Español": [3.9, 4.1, 4.5, 4.0],
        "Id": [123, 345, 567, None]
    }
)
```

Cambiar índices de fila

```python
notas = pd.DataFrame(
    {
        "Matematicas": [3.5, 4.1, 4.2, 3.2],
        "Español": [3.9, 4.1, 4.5, 4.0],
        "Id": [123, 345, 567, None]
    },
    index = [ f"Estudiante_{i}" for i in range(4)]
)
```

```python
notas
```

```python
notas = pd.DataFrame(
    {
        "Matematicas": [3.5, 4.1, 4.2, 3.2],
        "Español": [3.9, 4.1, 4.5, 4.0],
        "Id": [123, 345, 567, None]
    }
)
notas.set_index('Id', inplace = True)
```

```python
notas
```

Agregar una nueva columna

```python
notas["Artes"] = [4.8, 4.5, 4.6, 5.0]
```

```python
notas
```

Eliminar columnas

```python
notas.drop(columns="Artes", inplace=True)
```

Eliminar filas

```python
notas.drop(345, inplace=True)
```

```python
notas
```

Consultar información

```python
notas["Matematicas"]
```

```python
notas.Matematicas
```

Informacion pasajeros con exactamente 50 años

```python
data.loc[ data["Age"] == 50 ,["Name", "Age", "Sex"] ]
```

Diagrama de barras para comparar ¿ quienes sobrevivieron mas, hombres o mujeres ?

```python
d = data.loc[ data["Survived"] == 1, "Sex" ]
```

```python
d.describe()
```

Información de pasajeros con 50 años o más y de genero femenino que sobrevivieron

```python
data.loc[ (data["Age"] >= 50) & (data["Sex"] == "female") & (data["Survived"] == 1) ]
```

Información de los pasajeros mayores de edad que sobrevivieron

```python
data.loc[ (data["Age"] >= 18) & (data["Survived"] == 1)]
```

Pasajeros que embarcaron por la puerta "C" o "S"

```python
data["Embarked"].unique()
```

```python
data.loc[(data["Embarked"] == "C") | (data["Embarked"] == "S")]
```

Histograma de edades de los pasajeros que pagaron más de 400 libras o viajaban en primera clase

```python
data.loc[(data["Pclass"] == 1) | (data["Fare"]>400), "Age"].plot(kind="hist")
```

Información de pasajeros sin edad registrada

```python
data.loc[ data["Age"].isnull() ]
```

Información de pasajeros con edad registrada

```python
data.loc[ ~data["Age"].isnull() ]
```
