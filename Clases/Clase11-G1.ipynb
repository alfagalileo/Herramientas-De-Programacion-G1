# Herramientas de visualización

```python
# libreria para graficos estaticos
import matplotlib.pyplot as plt

# libreria para graficos dinamicos
import plotly.express as px

# libreria para dataframes
import pandas as pd
```

## Gráfico de líneas

```python
# primera forma
x = [1,2,3,4]
y = [-2,1,0,5]

# segunda forma
df = pd.DataFrame(
    {
        "x": [1,2,3,4],
        "y": [-2,1,0,5]
    }
)
```

```python
# grafica estatica
# plot -> grafica de lineas y puntos
plt.plot(x, y)
```

```python
plt.plot(x, y, color="r", marker="*", linestyle="-.", linewidth = 5, markersize = 15 )
```

Agregar varios datos a una misma gráfica

```python
plt.plot(x, y, color="r", marker="*", linestyle="-.", linewidth = 5, markersize = 15 )
plt.plot(y, x, color="g", marker="o", linestyle=":", linewidth = 2, markersize = 8 )
```

Gráfico completo

```python
# configurar el tamaño de la grafica
plt.figure(figsize = (6,3))

# crear los graficos
plt.plot( x, y, "-.r*", lw = 5, ms = 15, label = "Datos 1")
plt.plot( y, x, ":go", lw = 2, ms = 8, label = "Datos 2" )

# etiqueta de los ejes
plt.xlabel("Eje x (unidades x)")
plt.ylabel("Eje y (unidades y)")

# titulo de la figura
plt.title("Título de la figura")

# agregar una leyenda
plt.legend()

# mostrar la grafica
plt.show()
```

# Gráfico de funciones

```python
# libreria para calculos numericos
import numpy as np

x = np.linspace(0,10,200)
y = np.sin(x)
plt.plot(x,y)
```

Graficar entre $[0,15]$
- $\cos(x)$
- $\exp(-x^2)\times \sin(10 x)$

```python
x = np.linspace(0,15,500)
y1 = np.cos(x)
y2 = np.exp(-x**2)*np.sin(10*x)

plt.plot(x, y1, "b", lw=2, label = "cos(x)")
plt.plot(x, y2, "r", lw=2, label = "exp(-x^2)*sin(10*x)")
plt.legend()
plt.show()
```

# Gráfico de dispersión

```python
N = 1000

x = np.linspace(0,1,N)
y = np.random.rand(N)
c = np.random.rand(N)
s = np.random.rand(N)

plt.scatter(x,y, c = c, s = 120*s )
plt.show()
```

# Gráficos interactivos

```python
import plotly.graph_objects as go

N = 100
x = np.linspace(0,10, N)
y1 = np.random.rand(N) + 5
y2 = np.random.rand(N)
y3 = np.random.rand(N) - 5

plt.plot( x, y1)
plt.plot( x, y2)
plt.plot( x, y3)
plt.show()
```

```python
# crear una figura
fig = go.Figure()

# primera grafica
fig.add_trace(
    go.Scatter(x=x,y=y1, mode = "lines+markers", name = "datos1")
)

# segunda grafica
fig.add_trace(
    go.Scatter(x=x,y=y2, mode = "lines+markers", name = "datos2")
)

# segunda grafica
fig.add_trace(
    go.Scatter(x=x,y=y3, mode = "lines+markers", name = "datos2")
)
```
