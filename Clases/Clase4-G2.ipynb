## Operadores lógicos
Son operadores que permiten construir expresiones booleanas ( unir preguntas simples para obtener una respuesta a una pregunta múltiple)

- Operador `Y`

A ➡ pasar todas la materias

B ➡ estar a paz y salvo con la universidad

S ➡ resultado

| A       | B       | S       |
| ---     | ---     | ---     |
| `False` | `False` | `False` |
| `True`  | `False` | `False` |
| `False` | `True`  | `False` |
| `True`  | `True`  | `True`  |

- Operador `O`

_"Para finalizar mi carrera, hago una pasantía o una monografía"_

| A       | B       | S       |
| ---     | ---     | ---     |
| `False` | `False` | `False` |
| `True`  | `False` | `True`  |
| `False` | `True`  | `True`  |
| `True`  | `True`  | `True`  |

# Condicionales
Instrucciones para tomar decisiones dentro del código

```python
if (#condicion logica):
    # tarea a ejecutar en caso de que la pregunta sea cierta
else:
    # tarea a ejecutar en caso contrario
```

**Ej** Imprimir el mensajé `perdió` en caso de haber obtenido una nota inferior a 3.0. En caso contrario, imprimir `pasó`

````python
nota = 3.4

if (nota >= 3.0 ):
    print("pasó")
else:
    print("perdió")
````

````python
nota = 3.4

if (nota < 3.0 ):
    print("perdió")
else:
    print("pasó")
````

### Condicionales múltiples

```python
if (#condicion1):
    # tarea 1
elif (#condicion2):
    # tarea 2
elif (#condicion3):
    # tarea 3
else:
    # tarea por defecto
```

**Ej**

| letra | equivalencia |
| ---   | ---          |
| E     | (4, 5]       |
| S     | (3, 4]       |
| A     | (2, 3]       |
| I     | (1, 2]       |
| D     | [0, 1]       |

````python
def cambio_escala( nota ):
    if (nota > 4 ) and (nota <= 5 ):
        print("La nota es E")
    elif (nota > 3) and (nota <= 4):
        print("La nota es S")
    elif (nota > 2) and (nota <= 3):
        print("La nota es A")
    elif (nota > 1) and (nota <= 2):
        print("La nota es I")
    elif (nota >= 0) and (nota <= 1):
        print("La nota es D")
    else:
        print("Nota inválida")
````

````python
cambio_escala(4.8)
````

````python
initializeTurtle()
pendown()
speed(10)

def espiral(L: int, angulo: int):
    if (L < 1):
        print("Pare...")
    else:
        forward( L )
        right( angulo )
        espiral( L-2, angulo )

espiral( 100, 45 )
````

````python
initializeTurtle()
pendown()
speed(10)

def espiral2(L: int, angulo: int):
    if (L > 1):
        forward( L )
        right( angulo )
        espiral2( L-2, angulo )

espiral2( 100, 45 )
````

````python
initializeTurtle()
pendown()
speed(10)

def out_spiral(L: int, angulo: int):
    if (L < 200):
        forward( L )
        right( angulo )
        out_spiral( L+10, angulo )

out_spiral( 1, 90 )
````

````python
initializeTurtle()
pendown()
speed(10)

def angular_spiral(L: int, angulo: int):
    if ( angulo < 90 ):
        forward( L )
        right( angulo )
        angular_spiral( L-1, angulo+2 )

angular_spiral( 50, 10 )
````

**Ej**: Modelo de crecimiento de Solow

$
    K_{t+1} \approx s \times a \times \sqrt{K_t} + (1-d)\times K_t
$

donde $K_t$ es el capital actual, $s$ es la tasa de ahorro, $a$ es la productividad y $d$ es la tasa de depreciación.

````python
from math import sqrt

def solow( k, s, a, d, t):
    """
    k: capital incial
    s: tasa ahorro
    a: productividad
    d: tasa de depreciacion
    t: parametro de recurrencia (tiempo)
    """
    print( f"En el año t = {t} el capital es k = {k}" )

    if (t < 200):
        kn = s*a*sqrt(k) + (1-d)*k
        solow( kn, s, a, d, t+1 )
````

````python
solow(100, 10.0, 0.06, 0.1, 0 )
````
