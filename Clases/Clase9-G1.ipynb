# Lectura y escritura de archivos

## Creación y lectura de archivos

```python
f = open("nombre_de_archivo", "modificador")
```

Modificadores:

- `'r'` ➡ Abrir el archivo en modo lectura
- `'w', 'x'` ➡ Abrir el archivo en modo escritura. Borra el contenido anterior si el archivo existe.
- `'t'` ➡ Modo texto
- `'a'` ➡ Abrir el archivo en modo escritura agregando información al final del documento.

## Escritura de archivos

````python
# crear un archivo mis_datos.txt en modo escritura
f = open("mis_datos.txt", "w")
````

Escritura de información en el archivo

````python
f.writelines(["Pepito Pérez\t", "103567890\t", "3197293456\n"])
f.writelines(["Juanita Sarmiento\t", "10467890\t", "3225697823\t"])
# siempre cerrar el archivo para poder escribir
f.close()
````

Escritura iterativa

````python
dias = [ "M", "T", "W", "TH", "F" ]
gastos = [ 150_000, 100_000, 120_000, 50_000, 350_000 ]

with open("gastos_diarios.txt", "w") as f:
    for i in range(len(dias)):
        f.writelines([ dias[i], "\t", str(gastos[i]), "\n" ])
````

Lectura iterativa

````python
# primera forma: metodo read
with open("gastos_diarios.txt", "r") as g:
    # devuelve el contenido del archivo como una string
    print(g.read())

# segunda forma: metodo readlines
with open("gastos_diarios.txt", "r") as f:
    print(f.readlines())
````

````python
dias = []
gastos = []

with open("gastos_diarios.txt", "r") as f:
    lineas = f.readlines()

    for li in lineas:
        lineai = li.split("\t")
        dias.append( lineai[0] )
        gastos.append( int(lineai[1].strip()) )
````
