# Variables y expresiones

## Variables

Son espacios de memoria que permiten el almacenamiento de datos ( tipo de datos válidos ), bajo un nombre.

### Tipos de datos básicos

- Datos enteros (`int`). Ej `3, 5, ...`
     - para separar unidades de mil se puede usar (`_`)
- Datos decimales (`float`)
    - Para separar los decimales se usa (`.`) no (`,`)
- Cadenas de caracteres (`str`)
    - Se pueden definir usando comilla doble o comilla sencilla.

```python
# nota del taller 1
t1 = 4.6
t2 = 3.4
q1 = 3.9
q2 = 4.5
q3 = 2.8
q4 = 3.7
p = 3.6
```

```python
round( (t1 + t2)/2*0.33 + (q1 + q2 + q3 + q4)/4*0.33 + p*0.34, 1 )
```

Para definir una variable, usamos la siguiente regla

`NombreDeVariable = Valor`


Las variables sirven para evitar redundancias, consultar información previamente almacenada y realizar operaciones matemáticas en un formato más simple

*Ej*: Cálculo del interés simple

$
    I = C_0 i n
$

$C_0$ capital incial, $i$ es la tasa de interés, $n$ periodo de tiempo.

```python
# capital inicial
C0 = 12_000_000

# tasa de interes
i = 5/100

# periodo de tiempo
n = 10

I = C0*i*n
print( I )
```

```python
i = 0.07
I = C0*i*n
print( I )
```

# Nombramiento de variables

- Los nombres de variables deberían incluir únicamente caracteres alfa-numéricos (letras + números), guión bajo (`_`)
    - Caracteres prohibidos `,`, `"`. `'`, `#, &, >,<,?,¿,!,!,(,),{,},[,],^,.....`
- No se pueden dejar espácios. Ej ❌ Hoja 1 ➡ Hoja_1 ✔
- No se puede iniciar el nombre de una variable con un número. Ej ❌ 1er_caso ➡ caso_1 ✔
- No utilizar nombres de variables idénticos a las palabras claves del lenguaje.

```python
round = 12
```

```python
round(3.14,1)
```

# Reasignación de variables

Cada vez que se defina la misma variable con un dato distinto, el espacio de memoria asignado a esa variable toma el último valor que haya sido asignado a la variable. Este proceso se puede realizar con cualquier tipo de dato válido para el lenguaje

*Ej*: Interés compuesto

$
    C_n = C_0 (1 + i)^n
$

$C_n$ capital final, $C_0$ capital inicial, $i$ tasa de interés, $n$ periodo de tiempo


$C_0 = 35000000$, $i = 3 \% $ mensual, $n = 7$ años

```python
cap_inicial = 35_000_000
rate = 0.03
p = 7*12

capitalF = cap_inicial*( 1 + rate)**p

print("El capital final es: $", capitalF )
```

*Ej* una tortuga que se mueve sin aceleración en dirección rectilínea con velocidad $v = 2$ km/dia se detiene súbitamente al cabo de $40$ min. ¿ Qué distancia se desplazó la tortuga en ese tiempo ?

$
    x = v \times t
$


```python
velocidad_inicial = 2  # km/dia
t = 40/(24*60) # dias

x = velocidad_inicial*t

print( "El desplazamiento es: ", x, " km")
```

```python
velocidad_inicial = 2/(24*60)  # km/min
t = 40 # min

x = velocidad_inicial*t

print( "El desplazamiento es: ", x, " km")
```

*Ej* Calcular el PIB nominal de un país que tiene las siguientes características

- consumo: 500000
- inversión: 200000
- gasto público: 150000
- exportaciones: 80000
- importaciones: 100000

$
    PIB = \text{consumo} + \text{inversion} + \text{gasto publico} + \text{exportaciones} - \text{importaciones}
$

```python
consumo = 500_000
i = 200_000
gp = 150_000
exportaciones = 80_000
imp = 100_000

PIB = consumo + i + gp + exportaciones - imp
print("El PIB nominal del país es: ", PIB)
```

*Ej* Distancia entre dos puntos en el espacio euclídeo

$ P_1 = (1,5,2)$

$ P_2 = (-1, 10, 0)$

$
    d(P_1, P_2) = \sqrt{ (P_{1x} - P_{2x})^2 + (P_{1y} - P_{2y})^2  + (P_{1z} - P_{2z})^2  }
$


```python
p1x = 1
p1y = 5
p1z = 2

p2x = -1
p2y = 10
p2z = 0

import math
d = math.sqrt( (p1x - p2x)**2 + (p1y - p2y)**2 + (p1z - p2z)**2 )
print("La distancia entre los dos puntos es:", d, "m")
```
