## Operaciones sobre listas
Modificación de listas
- Agregar elementos a una lista previamente creada: `append`, `insert`, `extend`
- Eliminar elementos de una lista: `remove`, `pop`

### Método `append`
Sirve para agregar elementos a una lista siempre en la última posición

````python
a = [1,2]
print(a)
````

````python
a.append(3)
print(a)
````

**Ej**: Construcción iterativa de listas

`["*", "**", "***", ..., "*"*100]`

````python
# crear una lista vacia
l = []

n = 1
while n <= 10:
    l.append( "*"*n )
    n += 1
````

````python
l = []
for i in range(1,11):
    l.append("*"*i)
````

````python
l
````

**Ej**: Serie armónica

$$
    \sum_{k=1}^\infty \frac{1}{k}
$$

````python
# lista vacia
terminos = []

for k in range(1,10_000):
    terminos.append( 1/k )
````

````python
sum(terminos)
````

### Método `insert`
Agregar un elemento a la lista en una posición específica.

````python
l = ["*"*5, "*"*5, "*"*5]
l
````

````python
l.insert(2, "-"*5 )
l.insert(1, "-"*5 )
print(l)
````

### Método `extend`
Agrega los elementos de una lista `B` al final de una lista `A`

````python
A = [1,2,3,4]
B = [5,6,7,8]

A.extend(B)
````

````python
A
````

**Ej**

Usar el método `append` para crear dos listas con los patrones

```
   *
  * *
 * * *
* * * *
```

```
* * * *
 * * *
  * *
   *
```
Luego únalas con el método `extend`

````python
nmax = 10
l_arriba = []
for i in range(1,nmax+1):
    l_arriba.append( ("* "*i).center(2*nmax) )

l_abajo = []
for i in range(nmax-1, 0, -1):
    l_abajo.append( ("* "*i).center(2*nmax) )

l_arriba.extend(l_abajo)
````

````python
for li in l_arriba:
    print(li)
````

#### Regresión lineal simple

Dado un conjunto de datos $X$, $Y$ ($X$ son los datos independientes, $Y$ son los datos dependientes) con correlación lineal, la regresión lineal simple se expresa como:

$$
    Y = \beta_0 + \beta_1 X
$$

$$
    \beta_1 = \frac{ \overline{XY} - \overline{X} \overline{Y} }{ \overline{X^2} - \overline{X}^2 }
$$

$$
    \beta_0 = \overline{Y} - \beta_1 \overline{X}
$$
