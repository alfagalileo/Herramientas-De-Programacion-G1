# Creación de _Dahsboards_

Un dashboard es una herramienta de visualización (interactiva) para "contar historias" con la información. Estos procuran la facilidad en la comprensión del análisis de datos

## Elementos mínimos que conforman un Dashboard

- Conjunto de información
- Gráficos interactivos
- Tablas de datos (de ser relevantes)
- Widgets (botones, selectores, sliders .....)

## Plot

```python
pn.extension("plotly")

# libreria para graficos interactivos
import plotly.express as px

# libreria para dataframes
import pandas as pd

# crerar un dataframe
df = pd.DataFrame(
    {
        "Dias": ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"],
        "Gastos": [ 35_000, 450_000, 100_000, 50_000, 500_000 ]
    }
)

fig = px.line( df, x = "Dias", y = "Gastos")
pn.panel( fig )
```

## Distribución de elmentos gráficos en el Dashboard

```python
pn.extension()

pn.Column(
    pn.panel("# Dashboard de prueba", align="center"),
    pn.Row(
        pn.panel(fig, width=300, height=300),
        pn.Column(
            pn.panel("Mapa de Bogotá"),
            pn.pane.plot.Folium( mapa, width=200, height=200)
        ),
        c
    ),
)
```

## Interactividad (Reactividad)

```python
def enlace(evt):
    num.value = int(evt.new)

num = pn.indicators.Number( name = "Porcentaje", value = 0, format = "{value} %" )
sl = pn.widgets.FloatSlider( name = "valor", value = 0, start=0, end=100, step=1 )

# crear el callback
sl.param.watch( enlace, "value" )

dashboard = pn.Column(
    num,
    sl
)

dashboard.servable()
```

```python
sl1 =  pn.widgets.FloatSlider( name = "valor1", value = 0, start=0, end=100, step=1 )
sl2 =  pn.widgets.FloatSlider( name = "valor2", value = 0, start=0, end=100, step=1 )

def cambiar(evt):
    sl2.value = float(sl1.value)+5

sl1.param.watch( cambiar, "value")

ds = pn.Row(
    sl1,
    sl2
)

ds.servable()
```

## Conersor básico de divisas

```python
pn.extension()
```

```python
def conversion(evt):
    val = float(evt.new)*float(entrada.value)
    salida.value = str(val)

tasa = [
    ("USD -> EUR", "0.85"), ("EUR -> USD", "1.18"),
    ("USD -> COP", "3621"), ("EUR -> COP", "4259")
 ]

entrada = pn.widgets.TextInput( name = "Valor",  placeholder="Ingrese el valor a convertir")
selector = pn.widgets.MenuButton( name = "Transacción", items = tasa )
salida = pn.widgets.TextInput( name = "Resultado", disabled = True )

selector.on_click( conversion )

ds = pn.Column( entrada, selector, salida, height = 200)
ds.servable()
```
