## Expresiones regulares (`regex`)

Son patrones de búsqueda avanzados que permiten hallar información contenida dentro de una `str`.

Patrones de búsqueda básicos

- `[a]`  buscar por esos caracteres específicos.
- `[^ab]` buscar toos los caracteres distintos a `ab`.
- `[a-z]` buscar todos los caracteres en minúsculas.
- `[A-Z]` buscar todos los caracteres en mayúsculas.
- `\d` buscar dígitos.
- `\D` buscar todo lo que no sea un dígito.
- `\s` buscar espacios
- `\S` buscar todo lo que no sea un espacio.
- `\w` buscar caracteres (minúscula o mayúscula).
- `\W` buscar por cualquier cosa distinta a una letra.
- `+` modificador de agregación.

```python
# paquete para usar expresiones regulares
import re
```

## Función `findall`

Realiza la búsqueda de patrones expresados por medio de `regex` sobre una `str` y retorna todos los elementos que encontró

```python
txt = "Colombia es un gran País para vivir. 50 mill. de colombianos lo confirman. 01800-5666622 01800-34567"

re.findall( r"[A-Z]", txt )
```

```python
re.findall( r"[A-Z]\w+", txt )
```

```python
re.findall(r"\d+", txt )
```

```python
re.findall( r"01800-\d+", txt )
```

```python
re.findall( r"\d+-\d+", txt )
```

## Función `search`

## Funcion `sub`
Busca un patrón dentro de una string y la coincidencia la sustituye con otra string

```python
mi_variable = "1.) x = 1200 + y, es una igualdad matemática"
re.sub( r"\d+", "NÚMERO", mi_variable )
```

**Ej**: Identificación de códigos de factura

Extraer los códigos de las facturas de una empresa, con el siguiente patrón

`FAC-AÑO-NUMERO`

```python
txt = "Las facturas son: FAC-2023-12345, FAC-2024-010203 y FAC-2026-012345"
re.findall( r"FAC-\d+-\d+", txt )
```

**EJ:** Validación de números de cuenta
`###-###-###`

```python
def validacion( cuenta ):
    if re.search(r"\d{4}-\d{4}-\d{4}", cuenta ):
        print("Formato válido")
    else:
        print("Formato inválido")

cuenta = input("Ingrese su número de cuenta: ")
validacion( cuenta )
```
