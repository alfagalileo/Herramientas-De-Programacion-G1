# Mínimos Cuadrados

````python
def promedio(x: list):
    return sum(x)/len(x)

def minimos_cuadrados( x: list, y: list ):
    # calcular el promedio de x
    xm = promedio(x)
    ym = promedio(y)

    # crear una lista con los x^2
    x2 = [ x[i]**2 for i in range(len(x)) ]
    # calcular el promedio de x^2
    x2m = promedio(x2)

    xy = [ x[i]*y[i] for i in range(len(x))]
    xym = promedio(xy)

    # pendiente de la recta
    beta1 = (xym - xm*ym)/(x2m - xm**2)
    # punto de corte de la recta
    beta0 = ym - beta1*xm


    return [beta0, beta1]

inversion_publica = list(range(10,110,10))
PIB = [9.2, 2.4, 3.1, 4.3, 5.1, 6, 7.2, 7.9, 9, 10.2]

minimos_cuadrados( inversion_publica, PIB )
````

````python
# libreria para hacer graficos
import matplotlib.pyplot as plt
````

````python
# grafico de dispersion
plt.scatter( inversion_publica, PIB )
# mostrar la grafica
plt.show()
````

````python
inversion_publica = inversion_publica[1:]
PIB = PIB[1:]

res = minimos_cuadrados( inversion_publica, PIB)

y = [  res[0] + res[1]*inversion_publica[i] for i in range(len(PIB)) ]
plt.scatter( inversion_publica, PIB )
plt.plot( inversion_publica, y )
plt.show()
````

**Ej**: Tasas de cambio


````python
tasa = {
    ("USD", "EUR"): 0.86,
    ("EUR", "USD"): 1.16,
    ("COP", "USD"): 0.00027,
    ("USD", "COP"): 3717,
    ("EUR", "COP"): 4306,
}

transacciones = [
    ("USD", "EUR", 345),
    ("COP", "USD", 570_000),
    ("EUR", "COP", 123),
    ("EUR", "USD", 23)
]
````

````python
for ti in transacciones:
    cambio = ti[2]*tasa[ti[0],ti[1]]
    print(f"El cambio de {ti[2]} {ti[0]} es {cambio} {ti[1]}")
````
