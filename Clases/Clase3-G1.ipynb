# Funciones

![Imagen de funciones]( https://scaler.com/topics/images/what-is-a-function.webp )

Las funciones en un lenguaje de programación son porciones de código que se ejecutan a voluntad del usuario y permiten la factorización del código. De forma similar a las funciones matemáticas, las funciones tienen un conjunto de entradas (argumentos ó parámetros) y una salida.

**Ejemplo** cálculo del área superficial de un cilindro

$
    A(r,h) = 2 \pi r h
$

Partes de la función:
- Nombre de la función
- Parámetros o argumentos
- Cuerpo
- Salida

---
En Python, las funciones se declaran de la siguiente forma:

```python
def NombreDeFuncion( argumento1, argumento2, ... ):
    # cuerpo de la funcion
    return ValorOVarible
```


### Invocación de funciones
Para invocar una función se usa la siguiente regla:

```python
NombreDeFuncion( Valor1, Valor2, ... )
```

# Recomendaciones

⚡Los nombres de funciones y de los argumentos siguen las mismas reglas del nombramiento de variables.

⚡Las funciones pueden no depender de argumentos

⚡Después del `:` en la definición de la función, la siguiente línea debe iniciar con una tabulación.

```python
def NombreDeFuncion( ... ):
➡ #cuerpo de la funcion
➡ return Valor
```

⚡Existen funciones especiales que no retornan ningún valor. En ese caso no se necesita la instrucción `return`.

⚡Los argumentos de una función deben ser de un tipo de dato válido para Python: `int`, `float`, `str`, `bool`.


````python
from math import pi

# calculo del area de un cilíndro

def area_cilindro( r, h ):
    # cuerpo de la funcion
    a = round(2*pi*r*h, 2 )
    return a
````

````python
# invocar la funcion h = 2m, r = 40 cm
area_cilindro( 0.4, 2 )
````

````python
# area de un cilindro de h = 1.2 m, r = 1.8 m
radius = 1.8
height = 1.2
area_cilindro( radius, height )
````

````python
# area h = 4.7 m, r = 57 cm
r = 0.57
h = 4.7
a1 = area_cilindro( r, h )
print(f"El área del cilindro con r = {r} m y h = {h} m es: {a1} m^2")
````

**Ejemplo**: Cálculo del punto de equilibrio

$
    \text{PE} = \frac{ \text{CF}}{ \text{PU} - \text{CVU} }
$

- Caso 1: $CF = 12000$, $PU = 20000$, $CVU = 5000$

- Caso 2: $CF = 10000$, $PU = 15000$, $CVU = 5000$

- Caso 3: $CF = 18000$, $PU = 18000$, $CVU = 15000$

````python
import math as m

def punto_eq( cf: float, pu: float, cvu: float ):
    """
    Función para calcular el número de unidades requeridas para alcanzar
    el punto de equilibrio

    Argumentos:
        cf: costos fijos
        pu: precio por unidad
        cvu: costo variable por unidad

    Salida:
        Número de unidades
    """

    return m.ceil( cf/( pu - cvu) )
````

````python
punto_eq(12_000, 20_000, 5_000)
````

````python
punto_eq(10_000, 15_000, 5_000)
````

````python
punto_eq(18_000, 18_000, 5_000)
````

# Variables globales y locales

- Una variable local es aquella que se define dentro del cuerpo de una función y su valor solamente se puede usar al interior de la misma función.
- Una variable globarl es aquella cuyo valor se puede consultar en cualquier parte del código.

````python
# variable global
valor = 15

def func():
    # variable local
    n = 23
    print(n)

func()
````
